```python
import csv
import os
import pandas as pd
import numpy as np
from collections import defaultdict
from sklearn.model_selection import StratifiedGroupKFold
import matplotlib.pyplot as plt

# Path ke file metadata di Kaggle
METADATA_CSV_PATH = "/kaggle/input/notebooks/muhammadraiyansani/bdc-2-eda/extracted_full_image_metadata.csv"

# ==============================================================================
# 1. LOAD CSV MENGGUNAKAN PYTHON CSV READER (MEMBOBOTI BUG PANDAS PARSER)
# ==============================================================================
print("🚀 Memulai pemuatan metadata menggunakan Python CSV Reader...")
data = []
header = None

with open(METADATA_CSV_PATH, 'r', encoding='utf-8') as f:
    reader = csv.reader(f)
    header = next(reader)
    for row in reader:
        data.append(row)

df_meta = pd.DataFrame(data, columns=header)
print(f"✅ Berhasil memuat {len(df_meta)} baris data!")

# Konversi tipe data numerik
print("🔄 Mengonversi tipe data numerik...")
exclude_cols = ['file_path', 'filename', 'label', 'resolution_category', 'aspect_ratio_category', 'phash', 'dhash']
for col in df_meta.columns:
    if col not in exclude_cols:
        df_meta[col] = pd.to_numeric(df_meta[col], errors='coerce')

# ==============================================================================
# 2. PENGELOMPOKKAN GROUP ID & K-FOLD ON-THE-FLY (NUMPY & SKLEARN)
# ==============================================================================
df_clean = df_meta.copy()

print("⏳ Mengonversi pHash & dHash ke biner...")
def hex_series_to_binary_matrix(hex_series):
    hex_series = hex_series.fillna("0000000000000000")
    ints = hex_series.apply(lambda x: int(x, 16)).values.astype(np.uint64)
    shifts = np.arange(63, -1, -1, dtype=np.uint64)
    bits = (ints[:, None] >> shifts) & 1
    return np.where(bits, 1.0, -1.0).astype(np.float32)

H_p = hex_series_to_binary_matrix(df_clean['phash'])
H_d = hex_series_to_binary_matrix(df_clean['dhash'])

print("⏳ Menghitung jarak Hamming...")
D_p = (64 - np.dot(H_p, H_p.T)) / 2
D_d = (64 - np.dot(H_d, H_d.T)) / 2

# Menyaring indeks pasangan unik yang memiliki jarak pHash <= 4
idx_a, idx_b = np.where((D_p <= 4) & (np.triu(np.ones_like(D_p), k=1) > 0))

# Filter duplikat sejati dengan dHash <= 2
edges = []
for idx in range(len(idx_a)):
    i = idx_a[idx]
    j = idx_b[idx]
    d_dist = int(D_d[i, j])
    if d_dist <= 2:
        edges.append((i, j))

# Connected Components untuk menentukan ID kelompok duplikat
n = len(df_clean)
adj = defaultdict(list)
for u, v in edges:
    adj[u].append(v)
    adj[v].append(u)
    
visited = [False] * n
components = []
for i in range(n):
    if not visited[i]:
        component = []
        stack = [i]
        while stack:
            node = stack.pop()
            if not visited[node]:
                visited[node] = True
                component.append(node)
                for neighbor in adj[node]:
                    if not visited[neighbor]:
                        stack.append(neighbor)
        components.append(component)

# Koreksi label pada gambar konflik lintas kelas
print("🔧 Menyelaraskan label pada gambar konflik...")
for component in components:
    if len(component) > 1:
        labels = df_clean.iloc[component]['label'].values
        if len(set(labels)) > 1:
            filenames = df_clean.iloc[component]['filename'].values
            if "R_799.jpg" in filenames or "O_8873.jpg" in filenames:
                df_clean.loc[component, 'label'] = "0_Recyclable"
            elif "O_6678.jpg" in filenames or "67.waste-400x250-1.jpg" in filenames:
                df_clean.loc[component, 'label'] = "2_Organic"
            else:
                majority_label = df_clean.iloc[component]['label'].value_counts().idxmax()
                df_clean.loc[component, 'label'] = majority_label

group_ids = np.zeros(n, dtype=int)
for group_idx, component in enumerate(components):
    for node in component:
        group_ids[node] = group_idx
df_clean['group_id'] = group_ids

# Pembagian fold menggunakan Stratified Group K-Fold (5-Fold Split)
print("⚙️ Membagi K-Fold menggunakan Stratified Group K-Fold...")
sgkf = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=42)
df_clean['fold'] = -1
for fold, (train_idx, val_idx) in enumerate(sgkf.split(df_clean, y=df_clean['label'], groups=df_clean['group_id'])):
    df_clean.loc[val_idx, 'fold'] = fold

# ==============================================================================
# 3. STATISTIK PERBANDINGAN DAN DISTRIBUSI
# ==============================================================================
print("\n" + "="*50)
print("📊 RINGKASAN DATASET & PREPROCESSING")
print("="*50)
print(f"Total data gambar awal     : {len(df_meta)}")
print(f"Total data gambar akhir    : {len(df_clean)}")

# Perbandingan kelas sebelum vs sesudah koreksi label
print("\n🔄 DISTRIBUSI KELAS: SEBELUM VS SESUDAH KOREKSI LABEL")
dist_before = df_meta['label'].value_counts().sort_index()
dist_after = df_clean['label'].value_counts().sort_index()

df_dist_compare = pd.DataFrame({
    "Sebelum Koreksi": dist_before,
    "Sesudah Koreksi": dist_after,
    "Selisih Perubahan": dist_after - dist_before
})
display(df_dist_compare)

# Distribusi data per fold
print("\n📌 JUMLAH DATA PER KELAS DI TIAP K-FOLD (5-FOLD)")
fold_crosstab = pd.crosstab(df_clean['fold'], df_clean['label'], margins=True)
display(fold_crosstab)

# Persentase sebaran kelas per fold
print("\n📌 PERSENTASE SEBARAN KELAS DI TIAP K-FOLD (%)")
fold_crosstab_pct = pd.crosstab(df_clean['fold'], df_clean['label'], normalize='index') * 100
display(fold_crosstab_pct.round(2))

# ==============================================================================
# 4. RATA-RATA PROFIL VISUAL PER KELAS SAMPAH
# ==============================================================================
print("\n" + "="*50)
print("📊 RATA-RATA PROFIL VISUAL PER KELAS SAMPAH")
print("="*50)
cols_to_analyze = {
    'Lebar Gambar': ['width', 'width_pixels'],
    'Tinggi Gambar': ['height', 'height_pixels'],
    'Aspek Rasio': ['aspect_ratio'],
    'Kecerahan (Brightness)': ['mean_brightness', 'color_hsv_ch2_mean', 'moment_hsv_ch2_mean'],
    'Kontras (Contrast)': ['contrast', 'color_rgb_ch0_std', 'moment_rgb_ch0_std'],
    'Kerapatan Tepi (Canny)': ['edge_density'],
    'Keacakan Tekstur Mikro (LBP)': ['lbp_entropy'],
    'Energi Frekuensi Rendah (FFT)': ['fft_low_freq_energy'],
    'Energi Frekuensi Tinggi (FFT)': ['fft_high_freq_energy']
}

selected_features = {}
for name, options in cols_to_analyze.items():
    for opt in options:
        if opt in df_clean.columns:
            selected_features[name] = opt
            break

features_list = list(selected_features.values())
summary_mean = df_clean.groupby('label')[features_list].mean()

rename_dict = {v: k for k, v in selected_features.items()}
summary_mean = summary_mean.rename(columns=rename_dict)
display(summary_mean.round(4))

# ==============================================================================
# 5. VERIFIKASI DATA LEAKAGE (KEBOCORAN DATA)
# ==============================================================================
leakage_detected = False
for f in range(5):
    train_groups = set(df_clean[df_clean['fold'] != f]['group_id'])
    val_groups = set(df_clean[df_clean['fold'] == f]['group_id'])
    intersection = train_groups.intersection(val_groups)
    if len(intersection) > 0:
        leakage_detected = True
        print(f"⚠️ Peringatan: Kebocoran grup terdeteksi di Fold {f}! {len(intersection)} grup duplikat terpisah.")

if not leakage_detected:
    print("\n✅ VALIDASI LEAKAGE: 100% Bersih! Kelompok gambar duplikat tidak terpisah antar fold (Bebas Leakage).")
```

```python
import os
import csv
import joblib
import numpy as np
import pandas as pd
import cv2
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import f1_score, classification_report
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import timm
import albumentations as A
from albumentations.pytorch import ToTensorV2

# ==============================================================================
# 1. IDENTIFIKASI FITUR TABULAR TEKSTUR & FREKUENSI (BEBAS BIAS WARNA/KECERAHAN)
# ==============================================================================
exclude_cols = ['file_path', 'filename', 'label', 'resolution_category', 'aspect_ratio_category', 'phash', 'dhash', 'fold', 'group_id']

# Ambil semua kolom numerik
tabular_cols = [col for col in df_clean.columns if col not in exclude_cols]

# Bersihkan kolom yang mengandung kata kunci warna atau kecerahan
color_keywords = ['color', 'hue', 'saturation', 'brightness', 'rgb', 'hsv', 'lab']
tabular_cols = [col for col in tabular_cols if not any(kw in col.lower() for kw in color_keywords)]

print(f"✅ Berhasil menyaring {len(tabular_cols)} kolom fitur tekstur & frekuensi.")
print(f"Fitur terpilih meliputi: {tabular_cols[:10]} ... dst.")

# ==============================================================================
# 2. STANDARDIZATION & EKSPOR SCALER
# ==============================================================================
# Lakukan fitting scaler pada seluruh data train
scaler = StandardScaler()
df_clean[tabular_cols] = df_clean[tabular_cols].fillna(0)
scaler.fit(df_clean[tabular_cols].values)

# Simpan scaler ke file .pkl secara lokal untuk kebutuhan inferensi nanti
joblib.dump(scaler, 'tabular_scaler.pkl')
print("💾 Tabular Scaler berhasil disimpan ke: tabular_scaler.pkl")
```

```python
# ==============================================================================
# 1. PIPELINE AUGMENTASI ALBUMENTATIONS (AMAN GEOMETRIS)
# ==============================================================================
train_transforms = A.Compose([
    # Letterboxing untuk mengunci aspek rasio tanpa distorsi
    A.LongestMaxSize(max_size=224),
    A.PadIfNeeded(min_height=224, min_width=224, border_mode=cv2.BORDER_CONSTANT, value=0),
    
    # Transformasi geometri aman (tidak merubah nilai LBP/FFT/Contrast global)
    A.HorizontalFlip(p=0.5),
    A.VerticalFlip(p=0.5),
    A.RandomRotate90(p=0.5),
    A.ShiftScaleRotate(shift_limit=0.0625, scale_limit=0.1, rotate_limit=30, 
                       border_mode=cv2.BORDER_CONSTANT, value=0, p=0.5),
    
    A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
    ToTensorV2()
])

val_transforms = A.Compose([
    A.LongestMaxSize(max_size=224),
    A.PadIfNeeded(min_height=224, min_width=224, border_mode=cv2.BORDER_CONSTANT, value=0),
    A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
    ToTensorV2()
])

# ==============================================================================
# 2. DEFINISI CLASS PYTORCH DATASET
# ==============================================================================
class GarbageHybridDataset(Dataset):
    def __init__(self, df, tabular_cols, scaler, transforms=None):
        self.df = df.reset_index(drop=True)
        self.file_paths = self.df['file_path'].values
        
        # Transformasikan data tabular menggunakan scaler yang sudah fit
        self.tabular_data = scaler.transform(self.df[tabular_cols].values).astype(np.float32)
        
        # Ambil integer label dari nama folder (misal: "0_Recyclable" -> 0)
        self.labels = self.df['label'].apply(lambda x: int(x.split('_')[0])).values
        self.transforms = transforms

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        # Muat gambar
        img_path = self.file_paths[idx]
        image = cv2.imread(img_path)
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        
        # Terapkan augmentasi
        if self.transforms:
            augmented = self.transforms(image=image)
            image = augmented['image']
            
        tab_feat = torch.tensor(self.tabular_data[idx], dtype=torch.float32)
        label = torch.tensor(self.labels[idx], dtype=torch.long)
        
        return image, tab_feat, label
```

```python
# ==============================================================================
# 1. SPATIAL ATTENTION GATE (CBAM)
# ==============================================================================
class SpatialAttention(nn.Module):
    def __init__(self, kernel_size=7):
        super().__init__()
        padding = kernel_size // 2
        self.conv = nn.Conv2d(2, 1, kernel_size=kernel_size, padding=padding, bias=False)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        # Rata-rata dan Maksimum sepanjang dimensi Channel
        avg_out = torch.mean(x, dim=1, keepdim=True)
        max_out, _ = torch.max(x, dim=1, keepdim=True)
        x_out = torch.cat([avg_out, max_out], dim=1)
        x_out = self.conv(x_out)
        return x * self.sigmoid(x_out)

# ==============================================================================
# 2. FiLM MODULE (EARLY FUSION INJECTION)
# ==============================================================================
class FiLMModule(nn.Module):
    def __init__(self, num_tabular_features, num_channels):
        super().__init__()
        self.mlp = nn.Sequential(
            nn.Linear(num_tabular_features, 64),
            nn.ReLU(),
            nn.Linear(64, num_channels * 2)
        )
        self.num_channels = num_channels

    def forward(self, feature_map, tabular_features):
        film_params = self.mlp(tabular_features) # [B, C*2]
        gamma = film_params[:, :self.num_channels].unsqueeze(-1).unsqueeze(-1)
        beta = film_params[:, self.num_channels:].unsqueeze(-1).unsqueeze(-1)
        return (gamma * feature_map) + beta

# ==============================================================================
# 3. HYBRID CAFORMER-S18 MODEL
# ==============================================================================
class HybridCAFormerModel(nn.Module):
    def __init__(self, num_tabular_features, num_classes=3):
        super().__init__()
        # Load CAFormer S18 backbone
        self.backbone = timm.create_model('caformer_s18.sail_in22k_ft_in1k', pretrained=True, num_classes=0)
        self.num_features = 512 # Stage akhir caformer_s18 berukuran 512 channel
        
        # Modules
        self.film = FiLMModule(num_tabular_features, self.num_features)
        self.spatial_attention = SpatialAttention(kernel_size=7)
        
        # Main Classifier Head (Hasil Early Fusion)
        self.clf_joint = nn.Sequential(
            nn.Linear(self.num_features, 256),
            nn.LayerNorm(256),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(256, num_classes)
        )
        
        # Auxiliary Classifier Head (Gambar murni untuk memicu kemandirian visual)
        self.clf_img_aux = nn.Sequential(
            nn.Linear(self.num_features, 256),
            nn.LayerNorm(256),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(256, num_classes)
        )

    def forward(self, img, tab):
        # Ekstrak feature map lokal
        x_feat = self.backbone.forward_features(img)
        
        # Tangani format channels-last (B, H, W, C) jika dikembalikan oleh timm
        if x_feat.shape[-1] == self.num_features:
            x_feat = x_feat.permute(0, 3, 1, 2) # Ubah ke format (B, C, H, W)
            
        # 1. Cabang Auxiliary Image-Only (Pooling fitur bersih sebelum FiLM)
        x_img_pooled = torch.mean(x_feat, dim=[2, 3])
        out_img_aux = self.clf_img_aux(x_img_pooled)
        
        # 2. Cabang Utama (Early Fusion dengan FiLM)
        x_fused = self.film(x_feat, tab)
        
        # 3. Filter Atensi Spasial (Peredam Background)
        x_fused = self.spatial_attention(x_fused)
        
        # 4. Pooling Akhir & Klasifikasi
        x_fused_pooled = torch.mean(x_fused, dim=[2, 3])
        out_joint = self.clf_joint(x_fused_pooled)
        
        return out_joint, out_img_aux

    # Fungsi khusus mengekstrak Peta Atensi untuk Visualisasi Grad-CAM/Interpretability
    def get_attention_map(self, img, tab):
        self.eval()
        with torch.no_grad():
            x_feat = self.backbone.forward_features(img)
            if x_feat.shape[-1] == self.num_features:
                x_feat = x_feat.permute(0, 3, 1, 2)
            x_fused = self.film(x_feat, tab)
            
            # Hitung bobot sigmoid dari Spatial Attention
            avg_out = torch.mean(x_fused, dim=1, keepdim=True)
            max_out, _ = torch.max(x_fused, dim=1, keepdim=True)
            x_out = torch.cat([avg_out, max_out], dim=1)
            x_out = self.spatial_attention.conv(x_out)
            attention_mask = self.spatial_attention.sigmoid(x_out)
        return attention_mask # Output shape: [B, 1, H, W] (peta bobot atensi 2D)
```

```python
# Setup Device
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Running on device: {device}")

# Hitung Class Weights untuk mengatasi imbalance data
labels_int = df_clean['label'].apply(lambda x: int(x.split('_')[0])).values
weights = compute_class_weight('balanced', classes=np.unique(labels_int), y=labels_int)
class_weights = torch.tensor(weights, dtype=torch.float32).to(device)

print(f"Class Weights yang diterapkan: {weights}")

# Penampung Prediksi Out-Of-Fold (OOF)
oof_preds = np.zeros((len(df_clean), 3))
df_clean['pred_label'] = -1

# Hyperparameters
NUM_EPOCHS = 6
BATCH_SIZE = 64
LR = 5e-5

# ==============================================================================
# LOOP TRAINING 5-FOLD
# ==============================================================================
for fold in range(5):
    print("\n" + "="*70)
    print(f"🚀 MEMULAI PELATIHAN FOLD {fold} / 5")
    print("="*70)
    
    # Pisahkan Train dan Val berdasarkan fold
    train_df = df_clean[df_clean['fold'] != fold].reset_index(drop=True)
    val_df = df_clean[df_clean['fold'] == fold].reset_index(drop=True)
    val_indices = df_clean[df_clean['fold'] == fold].index
    
    # Inisialisasi Dataset & DataLoader
    train_dataset = GarbageHybridDataset(train_df, tabular_cols, scaler, transforms=train_transforms)
    val_dataset = GarbageHybridDataset(val_df, tabular_cols, scaler, transforms=val_transforms)
    
    train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=4, pin_memory=True)
    val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=4, pin_memory=True)
    
    # Inisialisasi Model, Optimizer, & Scheduler
    model = HybridCAFormerModel(num_tabular_features=len(tabular_cols), num_classes=3).to(device)
    optimizer = optim.AdamW(model.parameters(), lr=LR, weight_decay=1e-2)
    scheduler = optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=NUM_EPOCHS)
    
    criterion_joint = nn.CrossEntropyLoss(weight=class_weights)
    criterion_aux = nn.CrossEntropyLoss(weight=class_weights)
    
    best_val_f1 = 0.0
    
    # Loop Epoch
    for epoch in range(NUM_EPOCHS):
        # 1. TAHAP LATIHAN (TRAINING PHASE)
        model.train()
        train_loss = 0.0
        for images, tabs, labels in train_loader:
            images, tabs, labels = images.to(device), tabs.to(device), labels.to(device)
            
            optimizer.zero_grad()
            out_joint, out_aux = model(images, tabs)
            
            # Hitung Multi-Task Loss (1.0 untuk Image Aux Classifier)
            loss_joint = criterion_joint(out_joint, labels)
            loss_aux = criterion_aux(out_aux, labels)
            loss = loss_joint + 1.0 * loss_aux
            
            loss.backward()
            optimizer.step()
            train_loss += loss.item() * images.size(0)
            
        scheduler.step()
        train_loss /= len(train_loader.dataset)
        
        # 2. TAHAP VALIDASI (VALIDATION PHASE)
        model.eval()
        val_loss = 0.0
        all_preds = []
        all_labels = []
        all_probs = []
        
        with torch.no_grad():
            for images, tabs, labels in val_loader:
                images, tabs, labels = images.to(device), tabs.to(device), labels.to(device)
                out_joint, _ = model(images, tabs)
                
                loss_joint = criterion_joint(out_joint, labels)
                val_loss += loss_joint.item() * images.size(0)
                
                probs = torch.softmax(out_joint, dim=1)
                preds = torch.argmax(probs, dim=1)
                
                all_preds.extend(preds.cpu().numpy())
                all_labels.extend(labels.cpu().numpy())
                all_probs.extend(probs.cpu().numpy())
                
        val_loss /= len(val_loader.dataset)
        val_f1 = f1_score(all_labels, all_preds, average='macro')
        
        print(f"Epoch {epoch+1}/{NUM_EPOCHS} | Train Loss: {train_loss:.4f} | Val Loss: {val_loss:.4f} | Val F1-Macro: {val_f1:.4f}")
        
        # Simpan Model Terbaik Berdasarkan F1-Macro Tertinggi
        if val_f1 > best_val_f1:
            best_val_f1 = val_f1
            torch.save(model.state_dict(), f"best_model_fold_{fold}.pth")
            print(f"💾 Model Fold {fold} Terselamatkan! (F1-Macro: {best_val_f1:.4f})")
            
            # Catat prediksi OOF terbaik
            oof_preds[val_indices] = np.array(all_probs)
            
    print(f"✨ Selesai Fold {fold}. Best Val F1-Macro: {best_val_f1:.4f}")

# Simpan prediksi OOF ke dalam CSV
df_oof = pd.DataFrame(oof_preds, columns=['prob_0', 'prob_1', 'prob_2'])
df_oof['file_path'] = df_clean['file_path']
df_oof['label_asli'] = df_clean['label']
df_oof['pred_label'] = np.argmax(oof_preds, axis=1)
df_oof.to_csv('oof_predictions.csv', index=False)
print("\n🎉 OOF Predictions tersimpan ke: oof_predictions.csv")
```
